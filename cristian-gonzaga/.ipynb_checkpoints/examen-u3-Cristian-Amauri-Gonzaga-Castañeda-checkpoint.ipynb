```python
# # Parte 1. Conceptos
# # 1. Que es aprendizaje supervisado?
# Es un aprendizaje donde el modelo que va a aprender tiene los datos con sus respuestas conocidas de principio comparando con sus datos los valores
# # 2. Que es una variable objetivo?
# La variable objetivo es la cual va a compararse el resultado
# # 3. Que significa `venta_alta`?
# Una columna que nos clasifica el total de la venta, pero no numericamente
# # 4. Que diferencia hay entre `X` e `y`?
# Que X se utiliza como variables de entrada las cuales son las columnas del modelo con las que ira aprendiendo e y es la variable que tendrá para 
# predecir los datos
# # 5. Que es clasificacion?
# Es un aprendisaje que nos ayuda a juntar por así decir ciertos criterios que tenemos para las cosas
# # 6. Por que este problema es de clasificacion?
# Porque se usa el valor de la columna como clasificación, es decir, no requiere que sepamos el valor de la venta como tal (1500, 125, 345) si no más bien
# la o las reglas que sirven para clasificar según sea el caso calculado con el total de la venta (0, 1)
# # 7. Que hace `DecisionTreeClassifier`?
# Va tomando desiciones conforme avanzamos, es decir que toma desiciones de forma sistematica, y dependiendo de como vaya con sus casos, va llendose
# a las distintas ramas hasta llegar al final
# # 8. Para que sirve separar entrenamiento y prueba?
# Para separar los datos con los que el modelo aprendera y los datos con los que la prueba, separando la muestra de los datos
# # 9. Que significa exactitud?
# Significa saber que porcentaje del total de los datos fue correcto e incorrecto, entonces si tiene un 0.6 es un 60% de exactitud
# # 10. Para que sirve una matriz de confusion?
# Para saber con exactitud los resultados que salieron de los datos que se probaron, los verdaderos positivos, los verdaderos negativos,
# los falsos positivos y los falsos negativos
# # 11. Que significa `fit()`?
# Significa adaptar los datos para entrenar al modelo
# # 12. Que significa `predict()`?
# Significa predecir con los datos que le pasamos para que haga su predicción
# # 13. Por que se usa `pd.get_dummies()`?
# Para pasar las columnas que son textos a binarios, ya que el get_dummies lo que hace es una columna por valor unico
# # 14. Por que se usa `reindex()` al predecir ventas nuevas?
# Para forzar a despues de hacer el get_dummies que las columnas sean las mismas que se usaron al entrenar, entonces si una nueva columna cuando se hace
# el get_dummies aparece con el reindex se ignora
# # 15. Por que una prediccion no es una verdad absoluta?
# Porque depende de los datos que haya y puede haber casos donde sea erroneo y se marque como correcto, por eso debemos de tomarlo como una forma de 
# datos que nos ayuda a ver la manera y evolución de los mismos, sin confiar totalmente en ellos
```

```python
## Parte 2. Revision Del Dataset
```

```python
# 1. Muestra las primeras filas.
import pandas as pd
df = pd.read_csv("ventas_ecommerce_limpio.csv")
df.head()
```

```python
# 2. Revisa las columnas.
df.columns
```

```python
# 3. Muestra cuantas filas y columnas tiene.
df.shape
```

```python
# 4. Revisa si hay valores nulos.
df.isnull().sum()
```

```python
# 5. Verifica que exista la columna `total_venta`.
df.head()
# si existe
```

```python
# 6. Verifica que `total_venta` coincida con `cantidad * precio_unitario`.
venta_total = df["cantidad"] * df["precio_unitario"]
df["coincide"] = venta_total == df["total_venta"]
df.head()
df["coincide"].value_counts()
```

```python
# 7. Escribe una observacion breve sobre el estado del dataset.
# En primera instancia puedo observar que el dataset es un dataset sin valores nulos y con los datos correctamente alineados para el total de la venta
# que se obtiene con la cantidad de productos y precio unitario, también puedo ver que es una cantidad considerable de registros, no son muchos pero
# tampoco son pocos
```

```python
## Parte 3. Variable Objetivo
```

```python
df["venta_alta"] = df["total_venta"].apply(lambda x: 1 if x >= 1000 else 0)
df.head()
```

```python
df["venta_alta"].value_counts()
```

```python
# Por que venta_alta es la variable objetivo?
# Porque es la variable la cual es la respuesta de la regla que se puso para clasificar si es venta alta o si es venta no alta, sabiendo que es la
# respuesta es considerable ponerla como variable objetivo para que los datos del entrenamiento se comparen con esta misma
```

```python
## Parte 4. Variables De Entrada
```

```python
X = df[["cantidad", "precio_unitario", "categoria", "metodo_pago", "ciudad"]]
X.head()
```

```python
y = df["venta_alta"]
y.head()
```

```python
x = pd.get_dummies(X)
x.head()
```

```python
columnas_generadas = x.columns.to_list()
columnas_generadas
```

```python
# Por que no se debe usar total_venta como variable de entrada si venta_alta se creo a partir de total_venta?
# Porque es la respuesta a partir del total, lo que vendría siendo que le damos los datos con la respuesta y su equivalencia si es venta alta o es 
# venta baja
```

```python
## Parte 5. Entrenamiento Y Evaluacion
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
# 1. Divide los datos en entrenamiento y prueba.
# 2. Usa 80% entrenamiento y 20% prueba.
# 3. Usa `random_state=42`.

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
# 4. Entrena el modelo.
# 5. Genera predicciones con los datos de prueba.

modelo = DecisionTreeClassifier(random_state=42)
modelo.fit(X_train, y_train)
```

```python
# 6. Calcula exactitud.
# 7. Muestra matriz de confusion.

y_pred = modelo.predict(X_test)

from sklearn.metrics import accuracy_score, confusion_matrix

exactitud = accuracy_score(y_test, y_pred)
matriz = confusion_matrix(y_test, y_pred)

print(exactitud)
print(matriz)
```

```python
# 8. Crea una tabla llamada `resultados_prueba` con:

resultados_prueba = pd.DataFrame({
    "valor_real": y_test,
    "prediccion": y_pred
})

resultados_prueba["coincide"] = resultados_prueba["valor_real"] == resultados_prueba["prediccion"]

resultados_prueba
```

```python
resultados_prueba.value_counts()
```

```python
# 9. Cuenta cuantos aciertos y cuantos errores hubo.

aciertos = resultados_prueba[resultados_prueba["coincide"] == True ]
errores = resultados_prueba[resultados_prueba["coincide"] == False ]
```

```python
len(aciertos)
```

```python
len(errores)
```

```python
# # 1. Cual fue la exactitud?
# La exactitud fue de 1, quiere decir que la exactitud fue del 100%
# # 2. Cuantos aciertos tuvo el modelo?
# El modelo tuvo 12 aciertos
# # 3. Cuantos errores tuvo el modelo?
# El modelo tuvo 0 errores
# # 4. Que indica la matriz de confusion?
# Indica los verdaderos positivos, verdaderos negativos, los falsos positivos y falsos negativos
# # 5. Una buena exactitud significa que el modelo ya es perfecto? Explica.
# No, porque en este caso el dataset no tuvo casos donde hubiera más variedad con lo que esta en el csv, por ende si se usa en un futuro con muchos más
# datos puede que ahí ya presente las discrepancias
```
